```python
import pandas as pd
import json
```

```python
csv_path = r"C:\Users\chava\OneDrive\Desktop\UAM\Intel. de Negocios\3. Corte\Youtube\archive (1)\Youtube_data\Countries_data\USvideos.csv"
```

```python
df_videos = pd.read_csv(csv_path, delimiter=';')
```

```python
df_videos = df_videos.loc[:, ~df_videos.columns.str.contains('^Unnamed')]
```

```python
df_videos.head()
```

```python
df_videos.isnull().sum()
```

```python
df_videos = df_videos.dropna()
```

```python
df_videos = df_videos.dropna(how='all')
```

```python
df_videos = df_videos.fillna(0)
```

```python
df_videos.isnull().sum()
```

```python
json_path = r"C:\Users\chava\OneDrive\Desktop\UAM\Intel. de Negocios\3. Corte\Youtube\archive (1)\Youtube_data\Countries_data\US_category_id.json"
```

```python
with open(json_path, 'r') as f:
    category_data = json.load(f)
```

```python
categories = pd.json_normalize(category_data['items'], sep='_')
```

```python
print("Datos iniciales del JSON:")
print(categories.head())
```

```python
print("\nValores nulos en cada columna antes de limpiar:")
print(categories.isnull().sum())
```

```python
categories = categories.drop(columns=['kind', 'etag'])
```

```python
print("\nDatos después de eliminar columnas innecesarias:")
print(categories.head())
```
